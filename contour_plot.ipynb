```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Carga de datos
url = "https://raw.githubusercontent.com/plotly/datasets/master/volcano.csv"
data = pd.read_csv(url) 

# 2. Conviértelo en un array de NumPy
Z = data.values  
x = np.arange(Z.shape[1])  
y = np.arange(Z.shape[0]) 

# 3. Crea la malla para graficar
X, Y = np.meshgrid(x, y)

# 4. Genera el contour plot
plt.contourf(X, Y, Z, cmap='viridis') 
plt.title("Contour Plot - Volcano Dataset")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.colorbar(label="Elevación")
plt.show()
```
